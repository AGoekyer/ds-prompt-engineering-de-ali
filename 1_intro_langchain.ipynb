# Einführung in LangChain
LangChain ist ein beliebtes Framework, mit dem Sie schnell Anwendungen und Pipelines für Large Language Models (LLMs) erstellen können. Sie können damit Chatbots, RAGs, Agenten und vieles mehr entwickeln.

Die Grundidee der Bibliothek ist, dass wir eine _Chain_ (Kette) verschiedener Komponenten erstellen können, um komplexere Anwendungen zu entwickeln. Diese _Chains_ (Sie können sie sich als Pipelines vorstellen) können aus verschiedenen Komponenten bestehen, wie z.B.:
- **Prompt-Vorlagen**: Prompt-Vorlagen sind Vorlagen zur Generierung verschiedener Arten von Prompts, wie Chat-Prompts, Frage-Antwort-Prompts usw.
- **LLMs**: Large Language Models sind der Kern von LangChain. Sie können jedes LLM verwenden, das mit der Bibliothek kompatibel ist, wie OpenAI, Hugging Face, LLama usw.
- **Tools**: Tools sind Funktionen, die vom LLM verwendet werden können, um bestimmte Aufgaben auszuführen. Zum Beispiel können Sie ein Tool verwenden, um im Web zu suchen oder auf eine Datenbank zuzugreifen.
- **Agenten**: Agenten sind Komponenten, die LLMs und Tools verwenden können, um bestimmte Aufgaben auszuführen. Sie können verwendet werden, um Chatbots, **R**etrieval **A**ugmented **G**eneration (RAGs) usw. zu erstellen.
- **Retriever**: Retriever sind Komponenten, die verwendet werden können, um Informationen aus einer Datenbank oder Wissensbasis abzurufen. Sie können verwendet werden, um RAGs zu erstellen oder Informationen aus einer Datenbank abzurufen.
- **Memory**: Memory ist eine Komponente, die zum Speichern von Informationen über die Konversation verwendet werden kann. Sie kann eingesetzt werden, um Chatbots zu erstellen, die sich an frühere Gespräche erinnern können, oder um RAGs zu erstellen, die sich an frühere Abfragen erinnern können.

## Verwendung von LLMs in LangChain

LangChain unterstützt eine breite Palette von Anbietern für LLMs, darunter OpenAI, Hugging Face, Groq, LLama und viele andere.

Beginnen wir unsere Erkundung von LangChain mit der Groq-Integration.

### Groq-Integration
Groq ist ein Anbieter von LLMs, der leistungsstarke Inferenz-Fähigkeiten bietet. Um Groq mit LangChain zu verwenden, müssen Sie Ihren API-Schlüssel in der `.env`-Datei einrichten. Folgen Sie den Schritten in der README.md-Datei, um Ihre Umgebung einzurichten.

```python
from dotenv import load_dotenv
import warnings
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
```

#### Zugangsdaten aus der .env-Datei laden

```python
load_dotenv()
```

#### LLM definieren (mit Groq)

Wir können das LLM mit der [`ChatGroq`](https://python.langchain.com/docs/integrations/chat/groq/)-Klasse aus dem `langchain_groq`-Modul definieren.
Diese Klasse ermöglicht es uns, Folgendes festzulegen:
+ das Modell - unten verwenden wir `llama-3.1-8b-instant`
+ die Temperatur - wir setzen sie auf `0.1` für deterministischere Antworten
+ die maximale Token-Anzahl - wir setzen sie auf `512`, um die Antwortlänge zu begrenzen

```python
llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.1,
    max_tokens=512,
)
```

#### Prompt-Vorlage erstellen
Ein Prompt ist eine Reihe von Anweisungen oder Eingaben, die ein Nutzer einem LLM bereitstellt, um dessen Antwort zu steuern. Er hilft dem Modell, den Kontext zu verstehen und relevante Ausgaben zu generieren. In LangChain können wir eine Prompt-Vorlage mit der `PromptTemplate`-Klasse erstellen.

```python
template = """Question: {question}

Answer: """
prompt = PromptTemplate(template=template, input_variables=["question"])
```

Die __input_variables__ werden in der Vorlage mit geschweiften Klammern '{}' definiert. Dies ermöglicht es uns, dynamisch Werte in die Vorlage einzufügen, wenn wir sie verwenden.

#### Chain definieren
Eine Chain ist eine Abfolge von Komponenten, die nacheinander ausgeführt werden, um eine endgültige Ausgabe zu erzeugen. In LangChain können wir das Pipe-Symbol `|` verwenden, um eine Kette von Komponenten zu definieren. Die Ausgabe einer Komponente wird als Eingabe an die nächste Komponente in der Kette weitergegeben.

```python
chain = prompt | llm
```

#### Chain aufrufen

```python
question = "What is the backpropagation algorithm?"
```

```python
answer = chain.invoke(input={"question": question})
```

```python
print(answer.content.strip())
```

Wenn wir mehrere Fragen stellen möchten, können wir eine Liste von Dictionary-Objekten übergeben, wobei die Dictionaries die in unserer Prompt-Vorlage festgelegte Eingabevariable ("question") enthalten müssen, die auf die Frage abgebildet wird, die wir stellen möchten.

```python
qs = [ 
    {"question": "What is the backpropagation algorithm?"},
    {"question": "What is the purpose of the activation function in a neural network?"},
    {"question": "What is the difference between supervised and unsupervised learning?"},
    {"question": "Explain the concept of overfitting in machine learning."},
]
```

```python
answers = chain.batch(qs)
```

```python
for question, answer in zip(qs, answers):
    print("=" * 100)
    print(f"Question: {question['question']}")
    print(f"Answer: {answer.content.strip()}")
    print("=" * 100)
```
